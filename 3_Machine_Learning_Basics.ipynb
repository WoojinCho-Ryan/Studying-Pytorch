## 2. Hypothesis 수립
ML에서 식을 세울 때, 이 식을 Hypothesis라고 함.

일반적으로 Linear regression의 hypothesis는 y = Wx + b 또는 H(x) = Wx + b로 표기함. 여기서 W: Weight, b: bias.

## 3. Cost function에 대한 이해
cost function = loss function = error function = objective function
$cost(W, b) = \frac{1}{n} \sum_{i=1}^{n} [y^{(i)} - H(x^{(i)})]^2$를 minimize하는 W와 b가 training dataset를 가장 잘 나타내는 직선을 표현


## 4. Gradient Descent
**Optimizer** Algorithm - Gradient Descent
$$W := W - \alpha \frac{\partial}{\partial W} cost(W)$$

## 5. Pytorch로 linear regression implement

### 1) 기본 셋팅

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
```

```python
# 현재 실습하고 있는 python code를 재실행해도 다음에도 같은 결과가 나오도록 random seed 설정
torch.manual_seed(1)
```

### 3) Weight와 bias의 초기화

```python
# weight W를 0으로 초기화하고 학습을 통해 값이 변경되는 변수임을 명시함.
W = torch.zeros(1, requires_grad = True)
print(W)
```

```python
b = torch.zeros(1, requires_grad = True)
print(b)
```

### 4) Hypothesis 세우기
H(x) = Wx + b

### 5) Cost function 선언하기

```python
# 앞서 배운 torch.mean으로 평균을 구함.
cost = torch.mean((hypothesis - y_train) ** 2)
print(cost)
```

### 6) Gradient Descent implement

```python
# gradient를 0으로 초기화
optimizer.zero_grad()
# cost function을 derivative하여 gradient 계산
cost.backward()
# W와 b를 업데이트
optimizer.step()
```

7) 전체 코드

```python
# 데이터
x_train = torch.FloatTensor([[1], [2], [3]])
y_train = torch.FloatTensor([[2], [4], [6]])
# 모델 초기화
W = torch.zeros(1, requires_grad=True)
b = torch.zeros(1, requires_grad=True)
# optimizer 설정
optimizer = optim.SGD([W, b], lr=0.01)

nb_epochs = 2000 # 원하는만큼 Gradient Descent을 반복
for epoch in range(nb_epochs + 1):

    # H(x) 계산
    hypothesis = x_train * W + b

    # cost 계산
    cost = torch.mean((hypothesis - y_train) ** 2)

    # cost로 H(x) 개선
    optimizer.zero_grad()
    cost.backward()
    optimizer.step()

    # 100번마다 로그 출력
    if epoch % 100 == 0:
        print('Epoch {:4d}/{} W: {:.3f}, b: {:.3f} Cost: {:.6f}'.format(
            epoch, nb_epochs, W.item(), b.item(), cost.item()
        ))
```

Epoch: 전체 training dataset이 학습에 한 번 사용된 period. (여기서는 2000번)

실제 정답은 W=2, b=0인 것을 보아 거의 정답을 찾은 셈.

##6. optimizer.zero_grad()가 필요한 이유
PyTorch는 derivative을 통해 얻은 slope을 이전에 계산된 slope 값에 누적시키는 특징이 있음.

```python
import torch
w = torch.tensor(2.0, requires_grad = True)

nb_epochs = 20
for epoch in range(nb_epochs + 1):

  z = 2 * w

  z.backward()
  print('수식을 w로 미분한 값 : {}'.format(w.grad))
```

#3-2. Multivariable Linear regression
앞서 배운 $x$가 1개인 linear regression을 Simple Linear Regression이라고 한다. 여기서는 다수의 $x$로부터 $y$를 예측하는 Multivariable Linear Regression에 대하여 이해한다.

## 1. Data Definition
| Quiz 1 (x1) | Quiz 2 (x2) | Quiz 3 (x3) | Final (y) |
| :---: | :---: | :---: | :---: |
| 73 | 80 | 75 | 152 |
| 93 | 88 | 93 | 185 |
| 89 | 91 | 80 | 180 |
| 96 | 98 | 100 | 196 |
| 73 | 66 | 70 | 142 |

다음과 같은 training dataset이 있다. 이번에는 독립 변수 $x$의 개수가 3개라는 점을 참고하여, final score를 예측하는 model을 만들어보자.

$H(x) = w_{1}x_{1} + w_{2}x_{2} + w_{3}x_{3} + b$

## 2. Pytorch로 구현하기

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
```

```python
torch.manual_seed(1)
```

```python
# training set
x1_train = torch.FloatTensor([[73], [93], [89], [96], [73]])
x2_train = torch.FloatTensor([[80], [88], [91], [98], [66]])
x3_train = torch.FloatTensor([[75], [93], [90], [100], [70]])
y_train = torch.FloatTensor([[152], [185], [180], [196], [142]])
```

```python
# weight w와 bias b 초기화
w1 = torch.zeros(1, requires_grad = True)
w2 = torch.zeros(1, requires_grad = True)
w3 = torch.zeros(1, requires_grad = True)
b = torch.zeros(1, requires_grad = True)
```

```python
# optimizer 설정
optimizer = optim.SGD([w1, w2, w3, b], lr = 1e-5)

nb_epochs = 1000
for epoch in range(nb_epochs + 1):

  # H(x)
  hypothesis = x1_train * w1 + x2_train * w2 + x3_train * w3 + b

  # cost
  cost = torch.mean((hypothesis - y_train) ** 2)

  # cost로 H(x) 개선
  optimizer.zero_grad()
  cost.backward()
  optimizer.step()

  # 100번마다 로그 출력
  if epoch % 100 == 0:
        print('Epoch {:4d}/{} w1: {:.3f} w2: {:.3f} w3: {:.3f} b: {:.3f} Cost: {:.6f}'.format(
            epoch, nb_epochs, w1.item(), w2.item(), w3.item(), b.item(), cost.item()
        ))
```

## 3. Vector와 Matrix 연산으로 바꾸기

### 1) Vector 연산으로 이해하기
$H(X) = w_{1}x_{1} + w_{2}x_{2} + w_{3}x_{3}$는 아래와 같이 두 vector의 dot product로 표현 가능하다.

$$\begin{pmatrix} x_1 & x_2 & x_3 \end{pmatrix} \cdot \begin{pmatrix} w_1 \\ w_2 \\ w_3 \end{pmatrix} = (x_1 w_1 + x_2 w_2 + x_3 w_3)$$

두 vector를 각각 $X$, $W$로 표현하면 $H(X) = XW$로 쓸 수 있다.

### 2) Matrix 연산으로 이해하기

$$
\begin{pmatrix}
x_{11} & x_{12} & x_{13} \\
x_{21} & x_{22} & x_{23} \\
x_{31} & x_{32} & x_{33} \\
x_{41} & x_{42} & x_{43} \\
x_{51} & x_{52} & x_{53}
\end{pmatrix}
\begin{pmatrix}
w_{1} \\
w_{2} \\
w_{3}
\end{pmatrix}
+
\begin{pmatrix}
b \\
b \\
b \\
b \\
b
\end{pmatrix}
=
\begin{pmatrix}
x_{11}w_{1} + x_{12}w_{2} + x_{13}w_{3} + b \\
x_{21}w_{1} + x_{22}w_{2} + x_{23}w_{3} + b \\
x_{31}w_{1} + x_{32}w_{2} + x_{33}w_{3} + b \\
x_{41}w_{1} + x_{42}w_{2} + x_{43}w_{3} + b \\
x_{51}w_{1} + x_{52}w_{2} + x_{53}w_{3} + b
\end{pmatrix}
$$



$H(X) = XW + B$

## 4. Matrix 연산을 고려하여 PyTorch로 구현하기

```python
x_train = torch.FloatTensor([[73, 80, 75],
                             [93, 88, 93],
                             [89, 91, 80],
                             [96, 98, 100],
                             [73, 66, 70]])
y_train = torch.FloatTensor([[152], [185], [180], [196], [142]])
```

```python
print(x_train.shape)
print(y_train.shape)
```

```python
# Weight와 bias 선언
W = torch.zeros((3, 1), requires_grad = True)
b = torch.zeros(1, requires_grad = True)
```

여기서 주목할 점은 $W$의 크기가 (3 x 1) vector라는 점이다. 이는 matrix의 곱셈이 성립되기 위함이다.

```python
x_train  =  torch.FloatTensor([[73,  80,  75],
                               [93,  88,  93],
                               [89,  91,  80],
                               [96,  98,  100],
                               [73,  66,  70]])
y_train  =  torch.FloatTensor([[152],  [185],  [180],  [196],  [142]])

# 모델 초기화
W = torch.zeros((3, 1), requires_grad=True)
b = torch.zeros(1, requires_grad=True)
# optimizer 설정
optimizer = optim.SGD([W, b], lr=1e-5)

nb_epochs = 20
for epoch in range(nb_epochs + 1):

    # H(x) 계산
    # bias b는 broadcasting되어 각 샘플에 더해집니다.
    hypothesis = x_train.matmul(W) + b

    # cost 계산
    cost = torch.mean((hypothesis - y_train) ** 2)

    # cost로 H(x) 개선
    optimizer.zero_grad()
    cost.backward()
    optimizer.step()

    print('Epoch {:4d}/{} hypothesis: {} Cost: {:.6f}'.format(
        epoch, nb_epochs, hypothesis.squeeze().detach(), cost.item()
    ))
```

```python
# 임의의 입력 값에 대한 에측
with torch.no_grad(): # 이 블록 안에서 수행되는 모든 연산에 대해 역전파(즉, slope 계산)을 비활성화함.
  new_input = torch.FloatTensor([[75, 85, 72]]) # 예측하고 싶은 임의의 입력
  prediction = new_input.matmul(W) + b
  print('Predicted value for input {}: {}'.format(new_input.squeeze().tolist(), prediction.item()))
```
