```python
x_train = torch.FloatTensor([[1], [2], [3]])
y_train = torch.FloatTensor([[2], [4], [6]])
```

## 2. Hypothesis 수립
ML에서 식을 세울 때, 이 식을 Hypothesis라고 함.

일반적으로 Linear regression의 hypothesis는 y = Wx + b 또는 H(x) = Wx + b로 표기함. 여기서 W: Weight, b: bias.

## 3. Cost function에 대한 이해
cost function = loss function = error function = objective function
$cost(W, b) = \frac{1}{n} \sum_{i=1}^{n} [y^{(i)} - H(x^{(i)})]^2$를 minimize하는 W와 b가 training dataset를 가장 잘 나타내는 직선을 표현


## 4. Gradient Descent
**Optimizer** Algorithm - Gradient Descent
$$W := W - \alpha \frac{\partial}{\partial W} cost(W)$$

## 5. Pytorch로 linear regression implement

### 1) 기본 셋팅

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
```

```python
# 현재 실습하고 있는 python code를 재실행해도 다음에도 같은 결과가 나오도록 random seed 설정
torch.manual_seed(1)
```

### 2) 변수 선언

```python
x_train = torch.FloatTensor([[1], [2], [3]])
y_train = torch.FloatTensor([[2], [4], [6]])
```

```python
print(x_train)
print(x_train.shape)
```

```python
print(y_train)
print(y_train.shape)
```

### 3) Weight와 bias의 초기화

```python
# weight W를 0으로 초기화하고 학습을 통해 값이 변경되는 변수임을 명시함.
W = torch.zeros(1, requires_grad = True)
print(W)
```

```python
b = torch.zeros(1, requires_grad = True)
print(b)
```

### 4) Hypothesis 세우기
H(x) = Wx + b

```python
hypothesis = x_train * W + b
print(hypothesis)
```

### 5) Cost function 선언하기

```python
# 앞서 배운 torch.mean으로 평균을 구함.
cost = torch.mean((hypothesis - y_train) ** 2)
print(cost)
```

### 6) Gradient Descent implement

```python
# 'SGD'는 Gradient Descent의 일종. lr은 learning rate을 의미함.
optimizer = optim.SGD([W, b], lr = 0.01)
```

```python
# gradient를 0으로 초기화
optimizer.zero_grad()
# cost function을 derivative하여 gradient 계산
cost.backward()
# W와 b를 업데이트
optimizer.step()
```

7) 전체 코드

```python
# 데이터
x_train = torch.FloatTensor([[1], [2], [3]])
y_train = torch.FloatTensor([[2], [4], [6]])
# 모델 초기화
W = torch.zeros(1, requires_grad=True)
b = torch.zeros(1, requires_grad=True)
# optimizer 설정
optimizer = optim.SGD([W, b], lr=0.01)

nb_epochs = 2000 # 원하는만큼 Gradient Descent을 반복
for epoch in range(nb_epochs + 1):

    # H(x) 계산
    hypothesis = x_train * W + b

    # cost 계산
    cost = torch.mean((hypothesis - y_train) ** 2)

    # cost로 H(x) 개선
    optimizer.zero_grad()
    cost.backward()
    optimizer.step()

    # 100번마다 로그 출력
    if epoch % 100 == 0:
        print('Epoch {:4d}/{} W: {:.3f}, b: {:.3f} Cost: {:.6f}'.format(
            epoch, nb_epochs, W.item(), b.item(), cost.item()
        ))
```

Epoch: 전체 training dataset이 학습에 한 번 사용된 period. (여기서는 2000번)

실제 정답은 W=2, b=0인 것을 보아 거의 정답을 찾은 셈.

##6. optimizer.zero_grad()가 필요한 이유
PyTorch는 derivative을 통해 얻은 slope을 이전에 계산된 slope 값에 누적시키는 특징이 있음.

```python
import torch
w = torch.tensor(2.0, requires_grad = True)

nb_epochs = 20
for epoch in range(nb_epochs + 1):

  z = 2 * w

  z.backward()
  print('수식을 w로 미분한 값 : {}'.format(w.grad))
```
