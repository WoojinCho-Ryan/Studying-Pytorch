ML에서 식을 세울 때, 이 식을 Hypothesis라고 함.

일반적으로 Linear regression의 hypothesis는 y = Wx + b 또는 H(x) = Wx + b로 표기함. 여기서 W: Weight, b: bias.

## 3. Cost function에 대한 이해
cost function = loss function = error function = objective function
$cost(W, b) = \frac{1}{n} \sum_{i=1}^{n} [y^{(i)} - H(x^{(i)})]^2$를 minimize하는 W와 b가 training dataset를 가장 잘 나타내는 직선을 표현


## 4. Gradient Descent
**Optimizer** Algorithm - Gradient Descent
$$W := W - \alpha \frac{\partial}{\partial W} cost(W)$$

7) 전체 코드

```python
# 데이터
x_train = torch.FloatTensor([[1], [2], [3]])
y_train = torch.FloatTensor([[2], [4], [6]])
# 모델 초기화
W = torch.zeros(1, requires_grad=True)
b = torch.zeros(1, requires_grad=True)
# optimizer 설정
optimizer = optim.SGD([W, b], lr=0.01)

nb_epochs = 2000 # 원하는만큼 Gradient Descent을 반복
for epoch in range(nb_epochs + 1):

    # H(x) 계산
    hypothesis = x_train * W + b

    # cost 계산
    cost = torch.mean((hypothesis - y_train) ** 2)

    # cost로 H(x) 개선
    optimizer.zero_grad()
    cost.backward()
    optimizer.step()

    # 100번마다 로그 출력
    if epoch % 100 == 0:
        print('Epoch {:4d}/{} W: {:.3f}, b: {:.3f} Cost: {:.6f}'.format(
            epoch, nb_epochs, W.item(), b.item(), cost.item()
        ))
```

Epoch: 전체 training dataset이 학습에 한 번 사용된 period. (여기서는 2000번)

실제 정답은 W=2, b=0인 것을 보아 거의 정답을 찾은 셈.

#3-2. Multivariable Linear regression
앞서 배운 $x$가 1개인 linear regression을 Simple Linear Regression이라고 한다. 여기서는 다수의 $x$로부터 $y$를 예측하는 Multivariable Linear Regression에 대하여 이해한다.

## 1. Data Definition
| Quiz 1 (x1) | Quiz 2 (x2) | Quiz 3 (x3) | Final (y) |
| :---: | :---: | :---: | :---: |
| 73 | 80 | 75 | 152 |
| 93 | 88 | 93 | 185 |
| 89 | 91 | 80 | 180 |
| 96 | 98 | 100 | 196 |
| 73 | 66 | 70 | 142 |

다음과 같은 training dataset이 있다. 이번에는 독립 변수 $x$의 개수가 3개라는 점을 참고하여, final score를 예측하는 model을 만들어보자.

$H(x) = w_{1}x_{1} + w_{2}x_{2} + w_{3}x_{3} + b$

## 2. Pytorch로 구현하기

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
```

```python
torch.manual_seed(1)
```

```python
# training set
x1_train = torch.FloatTensor([[73], [93], [89], [96], [73]])
x2_train = torch.FloatTensor([[80], [88], [91], [98], [66]])
x3_train = torch.FloatTensor([[75], [93], [90], [100], [70]])
y_train = torch.FloatTensor([[152], [185], [180], [196], [142]])
```

```python
# weight w와 bias b 초기화
w1 = torch.zeros(1, requires_grad = True)
w2 = torch.zeros(1, requires_grad = True)
w3 = torch.zeros(1, requires_grad = True)
b = torch.zeros(1, requires_grad = True)
```

```python
# optimizer 설정
optimizer = optim.SGD([w1, w2, w3, b], lr = 1e-5)

nb_epochs = 1000
for epoch in range(nb_epochs + 1):

  # H(x)
  hypothesis = x1_train * w1 + x2_train * w2 + x3_train * w3 + b

  # cost
  cost = torch.mean((hypothesis - y_train) ** 2)

  # cost로 H(x) 개선
  optimizer.zero_grad()
  cost.backward()
  optimizer.step()

  # 100번마다 로그 출력
  if epoch % 100 == 0:
        print('Epoch {:4d}/{} w1: {:.3f} w2: {:.3f} w3: {:.3f} b: {:.3f} Cost: {:.6f}'.format(
            epoch, nb_epochs, w1.item(), w2.item(), w3.item(), b.item(), cost.item()
        ))
```

## 3. Vector와 Matrix 연산으로 바꾸기

### 1) Vector 연산으로 이해하기
$H(X) = w_{1}x_{1} + w_{2}x_{2} + w_{3}x_{3}$는 아래와 같이 두 vector의 dot product로 표현 가능하다.

$$\begin{pmatrix} x_1 & x_2 & x_3 \end{pmatrix} \cdot \begin{pmatrix} w_1 \\ w_2 \\ w_3 \end{pmatrix} = (x_1 w_1 + x_2 w_2 + x_3 w_3)$$

두 vector를 각각 $X$, $W$로 표현하면 $H(X) = XW$로 쓸 수 있다.

### 2) Matrix 연산으로 이해하기

$$
\begin{pmatrix}
x_{11} & x_{12} & x_{13} \\
x_{21} & x_{22} & x_{23} \\
x_{31} & x_{32} & x_{33} \\
x_{41} & x_{42} & x_{43} \\
x_{51} & x_{52} & x_{53}
\end{pmatrix}
\begin{pmatrix}
w_{1} \\
w_{2} \\
w_{3}
\end{pmatrix}
+
\begin{pmatrix}
b \\
b \\
b \\
b \\
b
\end{pmatrix}
=
\begin{pmatrix}
x_{11}w_{1} + x_{12}w_{2} + x_{13}w_{3} + b \\
x_{21}w_{1} + x_{22}w_{2} + x_{23}w_{3} + b \\
x_{31}w_{1} + x_{32}w_{2} + x_{33}w_{3} + b \\
x_{41}w_{1} + x_{42}w_{2} + x_{43}w_{3} + b \\
x_{51}w_{1} + x_{52}w_{2} + x_{53}w_{3} + b
\end{pmatrix}
$$



$H(X) = XW + B$

## 4. Matrix 연산을 고려하여 PyTorch로 구현하기

```python
x_train = torch.FloatTensor([[73, 80, 75],
                             [93, 88, 93],
                             [89, 91, 80],
                             [96, 98, 100],
                             [73, 66, 70]])
y_train = torch.FloatTensor([[152], [185], [180], [196], [142]])
```

```python
print(x_train.shape)
print(y_train.shape)
```

```python
# Weight와 bias 선언
W = torch.zeros((3, 1), requires_grad = True)
b = torch.zeros(1, requires_grad = True)
```

여기서 주목할 점은 $W$의 크기가 (3 x 1) vector라는 점이다. 이는 matrix의 곱셈이 성립되기 위함이다.

```python
x_train  =  torch.FloatTensor([[73,  80,  75],
                               [93,  88,  93],
                               [89,  91,  80],
                               [96,  98,  100],
                               [73,  66,  70]])
y_train  =  torch.FloatTensor([[152],  [185],  [180],  [196],  [142]])

# 모델 초기화
W = torch.zeros((3, 1), requires_grad=True)
b = torch.zeros(1, requires_grad=True)
# optimizer 설정
optimizer = optim.SGD([W, b], lr=1e-5)

nb_epochs = 20
for epoch in range(nb_epochs + 1):

    # H(x) 계산
    # bias b는 broadcasting되어 각 샘플에 더해집니다.
    hypothesis = x_train.matmul(W) + b

    # cost 계산
    cost = torch.mean((hypothesis - y_train) ** 2)

    # cost로 H(x) 개선
    optimizer.zero_grad()
    cost.backward()
    optimizer.step()

    print('Epoch {:4d}/{} hypothesis: {} Cost: {:.6f}'.format(
        epoch, nb_epochs, hypothesis.squeeze().detach(), cost.item()
    ))
```

```python
# 임의의 입력 값에 대한 에측
with torch.no_grad(): # 이 블록 안에서 수행되는 모든 연산에 대해 역전파(즉, slope 계산)을 비활성화함.
  new_input = torch.FloatTensor([[75, 85, 72]]) # 예측하고 싶은 임의의 입력
  prediction = new_input.matmul(W) + b
  print('Predicted value for input {}: {}'.format(new_input.squeeze().tolist(), prediction.item()))
```

예를 들면 PyTorch에서는 linear regression model이 nn.Linear()라는 함수로, 또 평균 제곱오차가 nn.functional.mse_loss()라는 함수로 구현되어 있음.

## 1. 단순 선형 회귀 구현하기

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
```

```python
torch.manual_seed(1)
```

이제 데이터를 선언함. 아래 데이터는 y=2x를 가정된 상태에서 만들어진 데이터로 우리는 이미 정답이 W = 2, b = 0임을 알고 있는 사태. 모델이 이 두 W와 b의 값을 제대로 찾아내도록 하는 것이 목표.

```python
# 데이터
x_train = torch.FloatTensor([[1], [2], [3]])
y_train = torch.FloatTensor([[2], [4], [6]])
```

```python
# 모델을 선언 및 초기화. 단순 선형 회귀이므로 input_dim = 1, output_dim = 1.
model = nn.Linear(1, 1)
```

model에는 weight W와 bias b가 저장되어져 있음. 이 값은 model.parameters()라는 함수를 사용하여 불러올 수 있음.

```python
# 전체 training dataset에 대해 Gradient Descent를 2000회 반복
nb_epochs = 2000
for epoch in range(nb_epochs + 1):

  # H(x) 계산
  prediction = model(x_train)

  # cost 계산
  cost = F.mse_loss(prediction, y_train) # Pytorch에서 제공하는 평균 제곱 오차 함수

  # cost로 H(x) 개선하는 부분
  # gradient를 0으로 초기화
  optimizer.zero_grad()
  # cost function을 미분하여 gradient 계산
  cost.backward()
  # W와 b를 업데이트
  optimizer.step()

  if epoch % 100 == 0:
    print('Epoch {:4d}/{} Cost: {:.6f}'.format(
          epoch, nb_epochs, cost.item()
      ))
```

여기서 다루는 내용은 linear regression에 한정되는 내용은 아님. 여기서는 데이터를 로드하는 방법과 Minibatch Gradient Descent에 대해 다룸.

## 1. Mini Batch and Batch Size

```python
x_train = torch.FloatTensor([[73, 80, 75],
                             [93, 88, 93],
                             [89, 91, 90],
                             [96, 98, 100],
                             [73, 66, 70]])
y_train = torch.FloatTensor([[152], [185], [180], [196], [142]])
```

만약 data의 개수가 매우 많다면 전체 데이터에 대해 gradient descent를 수행하는 것은 매우 느릴 뿐만 아니라 많은 계산량이 필요.

따라서 전체 data를 더 작은 단위로 나누어서 해당 단위로 학습하는 개념이 등장. 이 단위를 **Mini Batch**라고 함.

- Mini Batch 학습에서는 Mini Batch의 개수만큼 gradient descent를 수행해야 전체 data가 한 번 전부 사용되어 1 Epoch가 됨. 여기서 Mini Batch의 크기를 **batch size**라고 함.

- 전체 데이터에 대해서 한 번에 경사 하강법을 수행하는 방법을 '배치 경사 하강법'이라고 부름. 반면, 미니 배치 단위로 경사 하강법을 수행하는 방법을 '미니 배치 경사 하강법'이라고 부름.

- 배치 경사 하강법은 경사 하강법을 할 때, 전체 데이터를 사용하므로 가중치 값이 최적값에 수렴하는 과정이 매우 안정적이지만, 계산량이 너무 많이 필요. 미니 배치 경사 하강법은 경사 하강법을 할 때, 전체 데이터의 일부만을 보고 수행하므로 최적값으로 수렴하는 과정에서 값이 조금 헤매기도 하지만 훈련 속도가 빠름.

- 보통 batch size는 2의 제곱수를 사용. 그 이유는 CPU와 GPU의 메모리가 2의 배수이므로 데이터 송수신의 효율을 높일 수 있음.

## 4. Custom Dataset

torch.utils.data.Dataset을 상속받아서 직접 Custom Dataset을 만드는 경우도 있음. torch.utils.data.Dataset은 PyTorch에서 dataset을 제공하는 추상 클래스.

가장 기본적인 뼈대는 아래와 같음.

```python
class CustomDataset(torch.utils.data.Dataset):
  def __init__(self): # dataset의 전처리를 해주는 부분

  def __len__(self): # dataset의 길이. 즉, 총 sample의 수를 적어주는 부분

  def __getitem__(self, idx): # dataset에서 특정 1개의 sample을 가져오는 함수
```

## 5. Custom Dataset으로 linear regression 구현하기

```python
import torch
import torch.nn.functional as F
from torch.utils.data import Dataset
from torch.utils.data import DataLoader
```

```python
# Dataset 상속
class CustomDataset(Dataset):
  def __init__(self):
    self.x_data = [[73, 80, 75],
                   [93, 88, 93],
                   [89, 91, 90],
                   [96, 98, 100],
                   [73, 66, 70]]
    self.y_data = [[152], [185], [180], [196], [142]]

  # 총 data의 개수를 return
  def __len__(self):
    return len(self.x_data)

  # index를 입력받아 그에 맵핑되는 입출력 data를 PyTorch의 Tensor 형태로 return
  def __getitem__(self, idx):
    x = torch.FloatTensor(self.x_data[idx])
    y = torch.FloatTensor(self.y_data[idx])
    return x, y
```

```python
dataset = CustomDataset()
dataloader = DataLoader(dataset, batch_size = 2, shuffle = True)
```

```python
model = torch.nn.Linear(3, 1)
optimizer = torch.optim.SGD(model.parameters(), lr = 1e-5)
```

```python
nb_epochs = 20
for epoch in range(nb_epochs + 1):
  for batch_idx, samples in enumerate(dataloader):
    # print(batch_idx)
    # print(samples)
    x_train, y_train = samples
    # H(x) 계산
    prediction = model(x_train)

    # cost 계산
    cost = F.mse_loss(prediction, y_train)

    # cost로 H(x) 계산
    optimizer.zero_grad()
    cost.backward()
    optimizer.step()

    print('Epoch {:4d}/{} Batch {}/{} Cost: {:.6f}'.format(
        epoch, nb_epochs, batch_idx+1, len(dataloader),
        cost.item()
        ))
```

```python
# 임의의 입력 [73, 80, 75]를 선언
new_var =  torch.FloatTensor([[73, 80, 75]])
# 입력한 값 [73, 80, 75]에 대해서 예측값 y를 리턴받아서 pred_y에 저장
pred_y = model(new_var)
print("훈련 후 입력이 73, 80, 75일 때의 예측값 :", pred_y)
```
