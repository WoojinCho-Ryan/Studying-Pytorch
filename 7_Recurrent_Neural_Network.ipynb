# 7-1. Recurrent Neural Network, RNN

RNN: Sequence Model. input과 output을 sequence unit으로 처리.

## 1. Recurrent Neural Network, RNN

앞서 다룬 neural network는 hidden layer에서 activate function을 지난 값은 오직 output layer direction으로만 향함. 이러한 경우를 **Feed Forward Neural Network**라고 함.

RNN은 해당 경우는 아님. RNN은 hidden layer의 node에서 activate를 통해 나온 결과값을 output layer 방향으로도 보내면서, 다시 hidden layer node의 다음 계산의 input으로 보내는 특징을 가지고 있음.

image.png

$x$: input layer의 input vector, $y$: output layer의 output vector. bias $b$는 생략. RNN에서 hidden layer에서 activate function을 통해 결과를 내보내는 역할을 하는 node를 **cell**이라 함. 이 cell은 이전의 값을 기억하려고 하는 일종의 메모리 역할을 수행하므로 이를 **memory cell** 또는 **RNN cell**이라 함.

image.png

memory cell이 output layer 방향 또는 다음 시점 t+1의 자신에게 보내는 값을 **hidden state**라고 함.

image.png

input vector's dimension: 4

hidden state's size: 2

output layer의 output vector의 dimension: 2

RNN의 시점: 2

image.png

시점 t에서의 은닉 상태값: $h_t$

입력층에서 입력값을 위한 가중치: $W_x$

이전 시점 t-1의 은닉 상태값 $h_{t-1}$을 위한 가중치: $W_h$

- hidden layer
$h_t = tanh(W_xx_t+W_hh_{t-1}+b)$

- output layer
$y_t = f(W_yh_t+b)$

여기서 $f$는 nonlinear activate function 중 하나.
